```python
def check_sign(num):
    """Determines if a number is positive, negative, or zero."""
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    return "Zero"


def is_even(num):
    """Returns True if a number is even, False if odd."""
    return num % 2 == 0


def main():
    user_num = int(input("Enter a number: "))

    sign_status = check_sign(user_num)
    parity_status = "Even" if is_even(user_num) else "Odd"

    print(f"The number {user_num} is {sign_status} and {parity_status}.")


if __name__ == "__main__":
    main()
```
